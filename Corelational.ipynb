```python
variables = [
    "Physician_Density",
    "Hospital_Beds_Per_1000",
    "Health_Expenditure_y",
    "Out_of_Pocket",
    "Life_Expectancy",
    "Diabetes_Prevalence",
    "NCD_Mortality_Probability",
    "GDP_per_Capita",
    "Urban_Population"
]

corr_data = master[variables]
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix of Healthcare Accessibility and Health Outcomes")

plt.savefig("figures/correlation_heatmap.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

##  Correlation Matrix Analysis: Healthcare Accessibility & Health Outcomes

### Key Takeaways

#### 1. Healthcare Resources & Investment
* **Strong Positive Synergies:** `Physician_Density`, `Hospital_Beds_Per_1000`, and `Health_Expenditure_y` show extremely strong positive correlations ($r \ge 0.90$) with one another.
* **Economic Alignment:** Higher `GDP_per_Capita` strongly correlates with higher healthcare investment and provider availability ($r \ge 0.76$).

#### 2. Health Outcomes (`Life_Expectancy`)
* **Positive Drivers:** `Life_Expectancy` is strongly positively correlated with `GDP_per_Capita` (**0.88**), `Physician_Density` (**0.88**), and `Hospital_Beds_Per_1000` (**0.87**).
* **Major Negative Factor:** High `Out_of_Pocket` healthcare costs have a very strong inverse relationship with `Life_Expectancy` (**-0.98**).

#### 3. Disease Burden & Mortality Trends
* **Cost Burden & Mortality:** High `Out_of_Pocket` spending strongly correlates with higher `NCD_Mortality_Probability` (**0.92**) and increased `Diabetes_Prevalence` (**0.74**).
* **Urbanization Impact:** `Urban_Population` displays a near-perfect correlation with `Diabetes_Prevalence` (**0.98**) and `NCD_Mortality_Probability` (**0.92**), while inversely correlating with overall resource availability metrics.

---

> **Summary:** The data highlights that robust public/economic healthcare investment directly aligns with longer life expectancy, whereas reliance on out-of-pocket expenses and higher urbanization metrics strongly correlate with non-communicable disease risks and higher mortality rates.

```python
def scatter_plot(x, y, xlabel, ylabel, filename):

    plt.figure(figsize=(8,6))

    for country in ["Germany", "India"]:

        subset = master[master["Country"] == country]

        plt.scatter(
            subset[x],
            subset[y],
            s=50,
            label=country
        )

    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(f"{ylabel} vs {xlabel}")

    plt.legend()
    plt.grid(True)

    plt.savefig(f"figures/{filename}",
                dpi=300,
                bbox_inches="tight")

    plt.show()
```

```python
scatter_plot(
    "Physician_Density",
    "Life_Expectancy",
    "Physician Density",
    "Life Expectancy",
    "physicians_vs_life_expectancy.png"
)
```

##  Analysis: Life Expectancy vs. Physician Density (Germany vs. India)

###  Key Observations

* **Germany (Blue): High Density & Diminishing Returns**
  * **Positive Trend:** Strong positive relationship between physician density (2.7 to 4.5) and life expectancy (75 to 81 years).
  * **Plateau Effect:** As physician density passes **3.5 to 4.0**, life expectancy gains begin to plateau, demonstrating diminishing marginal returns on healthcare worker saturation.

* **India (Orange): Anomalous / Non-Linear Pattern**
  * **Clustered Density:** Physician density remains constrained within a low range (0.7 to 1.2).
  * **Negative Slope Artefact:** The plot exhibits a downward trend where higher reported physician density coincides with lower life expectancy (likely driven by historical time-series order, data reporting changes, or underlying confounding factors like population growth rates).

---

> **Takeaway:** Developing and developed nations operate in vastly different structural paradigms. High physician density supports longer life expectancy up to a saturation ceiling, whereas in low-density contexts, physician counts alone do not directly explain life expectancy trends without adjusting for broader socioeconomic metrics.

```python
scatter_plot(
    "Hospital_Beds_Per_1000",
    "Life_Expectancy",
    "Hospital Beds",
    "Life Expectancy",
    "beds_vs_life_expectancy.png"
)
```

##  Analysis: Life Expectancy vs. Hospital Beds (Germany vs. India)

###  Key Observations

* **Germany (Blue): Structural Reduction & Efficiency Gains**
  * **Inverse Trajectory:** As life expectancy increased from ~75 to over 81 years, hospital beds per capita decreased from ~10.4 down to ~7.5.
  * **Systemic Shift:** Reflects healthcare optimization (e.g., transition toward outpatient care, reduced length of stays, and better preventative management rather than relying on inpatient capacity).

* **India (Orange): Constrained Growth & Historical Shifts**
  * **Low Capacity Ceiling:** Hospital beds per capita remain tightly bound between ~0.5 and 2.1.
  * **Complex Trend:** Shows a two-phase historical path—an initial increase in beds alongside rising life expectancy, followed by a sharp vertical movement (life expectancy gains continuing even as bed capacity contracted/stabilized).

---

> **Takeaway:** Raw hospital bed count is not a simple linear indicator of population health. In advanced healthcare systems like Germany's, reducing bed density often signals improved efficiency rather than declining care, while developing systems show that broader public health improvements can drive life expectancy even under bed constraints.

## Analysis: Life Expectancy vs. Health Expenditure (Germany vs. India)

###  Key Observations

* **Germany (Blue): Strong Positive Returns & Diminishing Gains**
  * **Upward Trajectory:** Clear positive relationship between spending (from ~8.4% to over 12.5% of GDP) and life expectancy (rising from ~75 to ~81+ years).
  * **Diminishing Marginal Returns:** Higher spending initially yields steady gains in life expectancy, but the curve flattens above ~10.5% expenditure as it approaches human lifespan ceilings.

* **India (Orange): Paradoxical / Inverse Pattern**
  * **Low Spending Range:** Health expenditure remains tightly bounded within a lower window (~3.0% to 4.5% of GDP).
  * **Negative Slope Artefact:** The plot displays an inverse trend—as health expenditure metric values increase, recorded life expectancy appears lower. This typically indicates historical structural shifts (e.g., periods where economic crises or health shocks forced spending up as a percentage of GDP while life expectancy was lower or recovering).

---

> **Takeaway:** Sustained, high-level economic investment in national healthcare directly supports longer life expectancy up to a saturation point. In developing context trajectories, expenditure percentage alone does not linearly track life expectancy gains without accounting for real per-capita spending, baseline infrastructure, and time-series shocks.

##  Analysis: Life Expectancy vs. GDP per Capita (Germany vs. India)

###  Key Observations

* **India (Orange): Rapid Initial Gains**
  * **Steep Slope:** At lower income levels ($1,000 to $10,000 GDP per capita), small economic gains drive massive increases in life expectancy (rising sharply from ~58 to ~72 years).
  * **High Elasticity:** Reflects the dramatic health improvements that occur when basic needs—such as clean water, primary care, sanitation, and nutrition—are funded as a nation begins developing.

* **Germany (Blue): Logarithmic Plateau / Diminishing Returns**
  * **Flattening Curve:** At higher income levels ($20,000 to $70,000+ GDP per capita), life expectancy increases much more gradually (from ~75 to ~81 years) and eventually levels off.
  * **Preston Curve Dynamics:** Demonstrates classic economic diminishing returns—past a certain wealth threshold, additional increases in GDP yield smaller incremental gains in overall life expectancy.

---

> **Takeaway:** The plot illustrates a textbook **Preston Curve**: national wealth strongly drives health improvements in developing economies, but once basic healthcare infrastructure and standard of living are established, further increases in wealth produce diminishing returns on longevity.

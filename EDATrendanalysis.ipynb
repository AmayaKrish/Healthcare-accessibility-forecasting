```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Create figures folder if it doesn't exist
os.makedirs("figures", exist_ok=True)

# Load dataset
master = pd.read_csv(r"C:/Users/Admin/OneDrive/Documents/Thesis/master_dataset_updated(in).csv")
```

```python
plt.figure(figsize=(10,6))

for country in ["Germany", "India"]:
    data = master[master["Country"] == country]
    plt.plot(data["Year"],
             data["Physician_Density"],
             marker="o",
             linewidth=2,
             label=country)

plt.title("Physician Density in Germany and India (1990–2024)")
plt.xlabel("Year")
plt.ylabel("Physicians per 1,000 Population")
plt.legend()
plt.grid(True)

plt.savefig("figures/physician_density_trend.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

Germany (Blue): Shows a strong, steady upward trend, rising from ~2.75 physicians per 1,000 people in 1990 to roughly 4.5 by 2024.

India (Orange): Shows a gradual decline from 1.2 in 1990 down to below 1.0 around 2014, followed by a slight peak (~2019) and a plateau around 0.7 physicians per 1,000 people.

Key Takeaway: A widening gap in healthcare coverage per capita between the two nations over the 34-year period.

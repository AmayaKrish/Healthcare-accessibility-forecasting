```python
import pandas as pd

germany = pd.read_csv(r"C:/Users/Admin/OneDrive/Documents/Thesis/Germany_Healthcare_Data.csv")

germany_numeric = germany.select_dtypes(include=["number"])

germany_stats = germany_numeric.agg(["mean", "median", "std", "min", "max"]).T

germany_stats.columns = [
    "Mean",
    "Median",
    "Standard Deviation",
    "Minimum",
    "Maximum"
]

germany_stats = germany_stats.round(2)

display(germany_stats)

germany_stats.to_csv("Germany_Descriptive_Statistics.csv")
```

```python
import pandas as pd

germany = pd.read_csv(r"C:/Users/Admin/OneDrive/Documents/Thesis/Germany_Healthcare_Data.csv")
india = pd.read_csv(r"C:/Users/Admin/OneDrive/Documents/Thesis/India_Healthcare_Data.csv")

germany_mean = germany.mean(numeric_only=True)
india_mean = india.mean(numeric_only=True)

variables = [
    "Infant_Mortality",
    "GDP_per_Capita",
    "Urban_Population",
    "Total_Population",
    "Obesity_Prevalence",
    "Cardiovascular_Deaths_Rate",
    "Chronic_Respiratory_Deaths_Rate",
    "Diabetes_Deaths_Rate",
    "Neoplasms_Deaths_Rate",
    "UHC_Coverage_Index",
    "NCD_Mortality_Probability",
    "Diabetes_Prevalence",
    "Hospital_Density",
    "Life_Expectancy",
    "Out_of_Pocket_Payment",
    "Health_Expenditure",
    "Physician_Density",
    "Hospital_Beds_Per_1000",
    "Birth_Rate"
]

observations = []

for var in variables:
    g = germany_mean[var]
    i = india_mean[var]

    if g > i:
        observations.append(f"Germany has a higher average {var.replace('_',' ').lower()}.")
    elif g < i:
        observations.append(f"India has a higher average {var.replace('_',' ').lower()}.")
    else:
        observations.append("Both countries have similar averages.")

summary_table = pd.DataFrame({
    "Variable": variables,
    "Germany Mean": [round(germany_mean[v], 2) for v in variables],
    "India Mean": [round(india_mean[v], 2) for v in variables],
    "Initial Observation": observations
})


display(summary_table)

summary_table.to_csv("Variable_Summary_Table.csv", index=False)

print("Variable Summary Table saved successfully.")
```
